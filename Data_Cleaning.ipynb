```python
import pandas as pd
```

```python
pd.set_option("display.max_rows",None)
```

```python
lines = pd.read_csv("london.lines.csv")
stations = pd.read_csv("london.stations.csv")

# header = 1 , tells pandas to use the second row as column names
distance= pd.read_excel("Inter_station_database.xlsx",header= 1)
```

```python
lines.columns = lines.columns.str.strip()
```

```python
lines
```

```python
lines["name"].value_counts()
```

```python
lines["name"] = lines["name"].str.strip()
```

```python
lines.drop(columns = ["stripe"],inplace = True)
```

```python
# Removing the unwanted lines
lines = lines[(lines["name"] != "East London Line") &  (lines["name"] != "Docklands Light Railway")]
```

```python
lines["name"].value_counts()
```

```python
# Removing "Line" at the end to make it consisntent with the other dataset
lines["name"] = lines["name"].str.replace("Line", "").str.strip()
```

```python
lines["name"].value_counts()
```

```python
stations = stations.drop(columns = ["display_name"])
```

```python
distance.head()
```

```python
distance["Line"].value_counts()
```

```python
#Removing the unwanted lines that do not exist anymore
distance = distance[distance["Line"] != "East London"]
```

```python
distance["Line"].value_counts()
```

```python
# change the Line name to be consistent with the other dataset
distance.loc[distance["Line"] == "H & C", "Line"] = "Hammersmith & City"
```

```python
distance["Line"] = distance["Line"].str.replace("Line","").str.strip()
```

```python
all_stations = set(distance["Station from (A)"]) | set(distance["Station to (B)"])
```

```python
len(all_stations)
```

```python
distance.drop(columns = ["Un-impeded Running Time (Mins)"] , inplace = True)
distance.drop(columns = ["AM peak (0700-1000) Running Time (Mins)","Inter peak (1000 - 1600) Running time (mins)"], inplace = True)
```

```python
distance["Station from (A)"] = distance["Station from (A)"].str.lower()
distance["Station to (B)"] = distance["Station to (B)"].str.lower()
stations["name"] = stations["name"].str.lower()
```
