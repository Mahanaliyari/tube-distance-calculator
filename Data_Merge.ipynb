### Merge

```python
# merging to get the ids of "from" stations 
merged = distance.merge(stations[["id","name"]],how = "left",
                        left_on= "Station from (A)", right_on = "name").rename(columns = {"id": "from_id"}).drop(columns = "name")

# merging to get the ids of "to" stations 
merged = merged.merge(stations[["id","name"]],how = "left", 
                      left_on= "Station to (B)", right_on = "name").rename(columns = {"id":"to_id"}).drop(columns = "name")
```
